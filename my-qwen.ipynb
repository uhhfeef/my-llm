```python
import torch
import torch.nn as nn
```

```python
from torch.nn import LayerNorm, ReLU


class Transformer(nn.Module):
    def __init__(self, vocab_size, hidden, n_layers, n_heads, intermediate):
        super().__init__()

        self.embed_tokens = nn.Embedding(vocab_size, hidden)
        self.layers = nn.ModuleList()
        for _ in range(n_layers):
            self.layers.append(nn.ModuleDict({
                "attention": nn.MultiheadAttention(hidden, n_heads, dropout=0.0, bias=True,  batch_first=True)
                "norm1": nn.LayerNorm(hidden),
                "mlp": nn.Sequential(
                    nn.Linear(hidden, intermediate),
                    nn.ReLU(),
                    nn.Linear(intermediate, hidden)
                ),
                "norm2": nn.LayerNorm(hidden)
            }))
```
